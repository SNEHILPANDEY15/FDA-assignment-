```python
import pandas as pd
import sqlite3
import os

#Delete old database
if os.path.exists("bundesliga.db"):
    os.remove("bundesliga.db")

#Load dataset
df = pd.read_csv('/content/bulidata.csv')

#Remove index column
df = df.drop(columns=['Unnamed: 0'])

#Create new database
conn = sqlite3.connect('bundesliga.db')

#Store original dataset
df.to_sql('matches', conn, if_exists='replace', index=False)

print("Database created successfully.")
print(df.head())
```

```python
import sqlite3
import os

# Close any existing connection
try:
    conn.close()
except:
    pass

# Delete old database if it exists
if os.path.exists("bundesliga.db"):
    os.remove("bundesliga.db")

# Create a fresh database connection
conn = sqlite3.connect("bundesliga.db")
cursor = conn.cursor()

```

```python
teams = pd.concat([
    df[['HOME_TEAM_ID', 'HOME_TEAM_NAME']].rename(
        columns={'HOME_TEAM_ID':'TeamID',
                 'HOME_TEAM_NAME':'TeamName'}),
    df[['AWAY_TEAM_ID', 'AWAY_TEAM_NAME']].rename(
        columns={'AWAY_TEAM_ID':'TeamID',
                 'AWAY_TEAM_NAME':'TeamName'})
]).drop_duplicates()

teams.to_sql('Teams', conn, if_exists='replace', index=False)

display(pd.read_sql_query("SELECT * FROM Teams LIMIT 20", conn))
```

```python
seasons = df[['SEASON']].drop_duplicates().reset_index(drop=True)
seasons.index += 1
seasons = seasons.reset_index().rename(
    columns={'index':'SeasonID',
             'SEASON':'Season'})

seasons.to_sql('Seasons', conn, if_exists='replace', index=False)

display(pd.read_sql_query("SELECT * FROM Seasons", conn))
```

```python
locations = df[['LOCATION']].dropna().drop_duplicates().reset_index(drop=True)
locations.index += 1
locations = locations.reset_index().rename(
    columns={'index':'LocationID',
             'LOCATION':'Stadium'})

locations.to_sql('Locations', conn, if_exists='replace', index=False)

display(pd.read_sql_query("SELECT * FROM Locations LIMIT 20", conn))
```

```python
homewins = df[df['WIN_HOME'] == 1][[
    'HOME_TEAM_NAME',
    'AWAY_TEAM_NAME',
    'GOALS_HOME',
    'GOALS_AWAY'
]].copy()

homewins.index += 1
homewins = homewins.reset_index().rename(
    columns={'index':'MatchID',
             'HOME_TEAM_NAME':'HomeTeam',
             'AWAY_TEAM_NAME':'AwayTeam',
             'GOALS_HOME':'HomeGoals',
             'GOALS_AWAY':'AwayGoals'})

homewins.to_sql('HomeWins', conn, if_exists='replace', index=False)

display(pd.read_sql_query("SELECT * FROM HomeWins LIMIT 20", conn))
```

```python
goalstats = df[[
    'MATCH_DATE',
    'HOME_TEAM_NAME',
    'AWAY_TEAM_NAME',
    'GOALS_HOME',
    'GOALS_AWAY'
]].copy()

goalstats['TotalGoals'] = (
    goalstats['GOALS_HOME'] +
    goalstats['GOALS_AWAY']
)

goalstats.index += 1
goalstats = goalstats.reset_index().rename(
    columns={'index':'StatID',
             'MATCH_DATE':'MatchDate',
             'HOME_TEAM_NAME':'HomeTeam',
             'AWAY_TEAM_NAME':'AwayTeam'})

goalstats.to_sql('GoalStatistics', conn, if_exists='replace', index=False)

display(pd.read_sql_query(
    "SELECT * FROM GoalStatistics LIMIT 20",
    conn))
```

```python
display(pd.read_sql_query("""
SELECT HOME_TEAM_NAME,
       COUNT(*) AS HomeWins
FROM matches
WHERE WIN_HOME = 1
GROUP BY HOME_TEAM_NAME
ORDER BY HomeWins DESC
""", conn))
```

```python
display(pd.read_sql_query("""
SELECT SEASON,
       ROUND(AVG(GOALS_HOME + GOALS_AWAY),2)
       AS AvgGoals
FROM matches
GROUP BY SEASON
""", conn))
```

```python
display(pd.read_sql_query("""
SELECT t.TeamName,
       COUNT(h.MatchID) AS HomeWins
FROM Teams t
JOIN HomeWins h
ON t.TeamName = h.HomeTeam
GROUP BY t.TeamName
ORDER BY HomeWins DESC
LIMIT 20
""", conn))
```

```python
display(pd.read_sql_query("""
SELECT HOME_TEAM_NAME,
       COUNT(*) AS Wins,
       CASE
           WHEN COUNT(*) > 100 THEN 'High'
           WHEN COUNT(*) > 50 THEN 'Medium'
           ELSE 'Low'
       END AS Performance
FROM matches
WHERE WIN_HOME = 1
GROUP BY HOME_TEAM_NAME
ORDER BY Wins DESC
""", conn))
```

```python
display(pd.read_sql_query("""
SELECT HOME_TEAM_NAME,
       COUNT(*) AS Wins,
       RANK() OVER(
           ORDER BY COUNT(*) DESC
       ) AS TeamRank
FROM matches
WHERE WIN_HOME = 1
GROUP BY HOME_TEAM_NAME
""", conn))
```

```python
conn.execute("""
CREATE INDEX IF NOT EXISTS idx_team
ON matches(HOME_TEAM_NAME)
""")

display(pd.read_sql_query("""
SELECT HOME_TEAM_NAME,
       COUNT(*) AS MatchesPlayed
FROM matches
GROUP BY HOME_TEAM_NAME
ORDER BY MatchesPlayed DESC
LIMIT 20
""", conn))
```

```python
query = """
SELECT TEAM, SUM(WINS) AS TOTAL_WINS
FROM (
    SELECT HOME_TEAM_NAME AS TEAM,
           COUNT(*) AS WINS
    FROM matches
    WHERE WIN_HOME = 1
    GROUP BY HOME_TEAM_NAME

    UNION ALL

    SELECT AWAY_TEAM_NAME AS TEAM,
           COUNT(*) AS WINS
    FROM matches
    WHERE WIN_AWAY = 1
    GROUP BY AWAY_TEAM_NAME
)
GROUP BY TEAM
ORDER BY TOTAL_WINS DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT LOCATION,
       COUNT(*) AS MATCHES_PLAYED
FROM matches
GROUP BY LOCATION
ORDER BY MATCHES_PLAYED DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT SEASON,
       ROUND(AVG(DRAW) * 100, 2) AS DRAW_PERCENTAGE
FROM matches
GROUP BY SEASON
ORDER BY SEASON;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    MATCH_DATE,
    HOME_TEAM_NAME,
    AWAY_TEAM_NAME,
    GOALS_HOME + GOALS_AWAY AS TOTAL_GOALS,

    CASE
        WHEN GOALS_HOME + GOALS_AWAY >= 6 THEN 'High Scoring'
        WHEN GOALS_HOME + GOALS_AWAY >= 3 THEN 'Medium Scoring'
        ELSE 'Low Scoring'
    END AS MATCH_TYPE

FROM matches

LIMIT 20;
"""

pd.read_sql(query, conn)
```
